Around 500 CE, at the monastery of Nalanda, the mathematician Aryabhata laid the foundation for a 10-digit decimal system and around 628 CE Brahmagupta defined the concept of zero that was essential to the Hindu-Arabic system.

Five hundred years later, Leonardo of Pisa, known by his nickname Fibonacci, became the first to popularize the Hindu-Arabic system in Europe. He learned the system growing up in the Pisan trading post of Bejaia (in modern-day Algeria) and introduced it to Europe in 1202 when he published *Liber Abaci*, or the *Book of Calculation*. Because it was so much more efficient than the Roman numerals used at the time, it completely revolutionized commercial arithmetic and accounting.
