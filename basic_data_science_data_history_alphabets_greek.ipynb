By 800 B.C. the all-consonant Phoenician alphabet had made its way to [Greece](https://www.worldhistory.org/Greek_Alphabet/).
The Greeks kept some letters and removed others, but the primary change was to add letters as vowels to make it the first true alphabet
because they clarified how words are read and pronounced.
The early alphabets were read right to left, but the Greek alphabet was eventually read as we do today, left to right.

Different regional greek alphabets developed.
In 403 B.C. Athens officially decreed that only the variation from Ionia be used on the recommendation of Eucleides/
It was called the Euclidean alphabet, and it added the last vowels.
Byzantine scribes introduced lowercase letters about 850 A.D.
The Greek alphabet became the basis of all modern alphabets, including Latin that spread across Europe,
and Cyrillic that became the basis of the Russian alphabet.
