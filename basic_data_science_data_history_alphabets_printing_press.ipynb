## The Printing Press

It laid the groundwork for creating a method to quickly spread writing to a wide audience,
, creating a unified visual language. Centuries later, when early Italian printers were looking for a clean font style to cast in metal for the printing press, they mistook Carolingian minuscule for classical Roman handwriting. They digitized it into the "Roman" and "Italic" typefaces we still read on our screens and in books today.
