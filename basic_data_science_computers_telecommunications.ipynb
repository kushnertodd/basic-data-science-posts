Telecommunications character encoding converts human-readable letters, numbers, and symbols into binary or discrete electrical signals for transmission across telegraph, telephone, mobile, and computer networks. [1, 2, 3]
Major Telecommunication Character Encodings
Telegraph and Early Serial Codes
Baudot Code (ITA1 / ITA2): A 5-bit stateful encoding system developed by Émile Baudot in the 1870s and standardized as International Telegraph Alphabet No. 2 (ITA2). It used "letters" and "figures" shift characters to map more than 32 symbols across mechanical teletypewriters, Telex, and RTTY. [4, 5]
Computer and Wireline Standards
ASCII (American Standard Code for Information Interchange): A 7-bit standard established in the 1960s that encodes 128 characters, including basic Latin letters (A–Z, a–z), digits (0–9), and control signals. [2, 6, 7]
Extended ASCII / Latin-1 (ISO/IEC 8859-1): An 8-bit extension that adds 128 additional code points to support Western European diacritics and symbols. [4, 7]
ISO/IEC 6937: A multi-byte or coded character set designed for text communication across diverse European multi-lingual applications. [8]
Mobile Telephony and Wireless Messaging
GSM 03.38 (GSM-7): A 7-bit default alphabet mandatory for GSM handsets and network elements, used for SMS, Cell Broadcast, and USSD. A standard single SMS holds 160 characters using GSM-7 (140 bytes). [9, 10, 11]
UCS-2 (16-bit Universal Coded Character Set): Used in mobile messaging when non-Western scripts (such as Chinese, Korean, or Japanese) or emojis are included, which cuts the single SMS character limit down to 70 characters. [10, 11]
Modern Global Standard
Unicode / UTF-8: A variable-width (8-bit to 32-bit) universal character encoding that is fully backward-compatible with 7-bit ASCII. It serves as the core text format across the internet, VoIP, and modern digital data streams. [4, 12]
