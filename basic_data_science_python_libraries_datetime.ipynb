## [Date and Time functions](https://docs.python.org/3/library/datetime.html) - `datetime` library
The `datetime` module allows handling dates, times, and durations.
These are handled with a number of datetime `objects`:
- `date`: calendar dates
- `time`: date with time of day
- `timedelta`: durations

```python
# today's date
import datetime
datetime.date.today()
```

```python
# an arbitrary date
import datetime
datetime.date(2020, 12, 31)
```

```python
# replace the day in a date
import datetime
datetime.date(2020, 12, 31).replace(day=1)
```

```python
# replace the month in a date
import datetime
datetime.date(2020, 12, 31).replace(month=8)
```

```python
# replace the year in a date
import datetime
datetime.date(2020, 12, 31).replace(year=2015)
```

```python
# get the weekeday of today's date (as a number)
import datetime
datetime.date.today().weekday()
```

```python
# get the current time in 24-hour format (with milliseconds)
import datetime
datetime.datetime.now().time()
```

```python
# get the current hour
import datetime
datetime.datetime.now().time().hour
```

```python
# get the current minute
import datetime
datetime.datetime.now().time().minute
```

```python
# get a duration (as days and seconds)
import datetime
datetime.timedelta(
    days=50,
    seconds=27,
    minutes=5,
    hours=8,
    weeks=2
)
```

```python
# get a duration as the difference between two times
import datetime
datetime.datetime(2026, 8, 16, 10, 0, 0) - datetime.datetime(2026, 8, 14, 12, 30, 45)
```

The `datetime` library has functions with a fairly extensive range of options for converting dates to text with `strftime`
and converting text to dates with `strptime`.

| Directive | Meaning | Example |
| - | - | - |
| %a | Weekday abbreviated name. | Sun, Mon, ..., Sat |
| %A | Weekday full name | Sunday, Monday, ..., Saturday |
| %w | Weekday as a decimal number, where 0 is Sunday and 6 is Saturday | 0, 1, ..., 6 |
| %d | Day of the month as a zero-padded decimal number | 01, 02, ..., 31 |
| %b | Month abbreviated name | Jan, Feb, ..., Dec |
| %B | Month full name | January, February, ..., December |
| %m | Month as a zero-padded decimal number | 01, 02, ..., 12 |
| %y | Year without century as a zero-padded decimal number | 00, 01, ..., 99 |
| %Y | Year with century as a decimal number | 0001, 0002, ..., 2013, 2014, ..., 9998, 9999 |
| %H | Hour (24-hour clock) as a zero-padded decimal number | 00, 01, ..., 23 |
| %I | Hour (12-hour clock) as a zero-padded decimal number | 01, 02, ..., 12 |
| %p | AM or PM | AM, PM |
| %M | Minute as a zero-padded decimal number | 00, 01, ..., 59 |
| %S | Second as a zero-padded decimal number | 00, 01, ..., 59 |
| %f | Microsecond as a decimal number, zero-padded to 6 digits | 000000, 000001, ..., 999999 |
| %j | Day of the year as a zero-padded decimal number | 001, 002, ..., 366 |
| %c | Full date and time  | Tue Aug 16 21:30:00 1988 |
| %x | Full date | 08/16/1988 |
| %X | Full time (24 hour) | 21:30:00 |
| %% | A literal '%'character | % |

```python
# format today as a full date/time
import datetime
datetime.datetime.now().strftime("%c")
```

```python
# format today date/time like "05/30/2022 04:12:50 AM"
import datetime
datetime.datetime.now().strftime("%m/%d/%Y %I:%M:%S %p")
```

```python
# format today date like "09/12/2022"
import datetime
datetime.datetime.now().strftime("%m/%d/%Y")
```

```python
# format today date/time like "October 5, 2022"
import datetime
datetime.datetime.now().strftime("%B %d, %Y")
```

```python
# format today date/time like "04-Feb-2022" with 12-hour time
import datetime
datetime.datetime.now().strftime("%d-%b-%Y")
```

```python
# Translate today date/time like "05/30/2022 04:12:50 AM" to datetime object
import datetime
datetime.datetime.strptime("05/30/2022 04:12:50 AM","%m/%d/%Y %I:%M:%S %p")
```

```python
# Translate today date/time like October 5, 2022" to datetime object
import datetime
datetime.datetime.strptime("October 5, 2022", "%B %d, %Y")
```
