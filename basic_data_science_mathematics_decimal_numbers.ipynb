# Decimal numeral system

In 269 BC, [Ashoka the Great](https://worldhistoryedu.com/ashoka-life-and-major-accomplishments-of-the-3rd-mauryan-emperor/) became king of the [Mauryan Empire](https://education.nationalgeographic.org/resource/mauryan-empire/) that covered most of the Indian subcontinent.
He created dozens of [pillars](https://www.worldhistory.org/Edicts_of_Ashoka/)
with inscriptions of [brahmi script](https://en.wikipedia.org/wiki/Brahmi_script) that included the first instances of [decimal numerals](https://en.wikipedia.org/wiki/Brahmi_numerals).
Around the third or fourth century CE, the symbol for zero as an empty value first appeared as a dot in a document, the [Bakhshali manuscript](https://www.glam.ox.ac.uk/article/carbon-dating-finds-bakhshali-manuscript-contains-oldest-recorded-origins-symbol-zero).

In 427 CE Emperor Kumaragupta I of the Gupta Empire founded the monestary of [Nalanda](https://scroll.in/article/1077528/a-new-book-shows-how-nalanda-mahavihara-contributed-to-the-fields-of-ancient-mathematics-astronomy).
There, around 500 AD, the mathematician [Aryabhata](https://cheenta.com/how-did-aryabhatta-invent-zero-how-did-he-get-this-idea-why-did-he-give-zero-an-oval-shape/)
described how a placeholder like zero would be used in a decimal place value system.

In 628 CE, [Brahmgupta](https://www.open.ac.uk/blogs/MathEd/index.php/2022/08/25/the-men-who-invented-zero/)
was the first to realize zero was needed as a number just like the other nine for the decimal system, rather than merely as an absent value, and developed rules for using zero with other numbers.
His work was translated into Arabic and from it, around 825 CE the
mathematician [Muhammad ibn Musa Khwarizmi](https://www.historyofinformation.com/detail.php?id=202) (780-850) wrote an explanation of the use of the zero in the decimal system using Indian numerals.

Five hundred years later, Leonardo of Pisa, known by his nickname [Fibonacci](https://library.ethz.ch/en/collections-and-archives/platforms/virtual-exhibitions/fibonacci-un-ponte-sul-mediterraneo/biography-of-leonardo-of-pisa.html), had grown up in a Pisan trading post in Bejaia,
There he learned the decimal place value system and in 1202 wrote the
[Liber Abaci](https://historyofinformation.com/detail.php?entryid=278), the 'Book of Calculation', and was the first to popularize in Europe the use of the Hindu-Arabic numeral system.
It was so much more efficient than the Roman numerals used in Europe for commercial arithmetic at the time that it revolutionized accounting.
