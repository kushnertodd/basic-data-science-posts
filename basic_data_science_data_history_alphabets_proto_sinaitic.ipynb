## The Proto-Sinaitic Alphabet

Over four thousand years ago pharaohs sent expeditions to the Sinai Peninsula in search of turquoise,
which was prized by the Egyptian court for protection and allowing rulers to pass to the afterlife.
On a plateau in south-west Sinai was [Serabit el-Khadim](https://www.penn.museum/sites/expedition/the-alphabet/),
the most important site of mines abundant in turquoise.
Semitic-speaking laborers from the land of Canaan were imported to work in the mines.
Around 1900 B.C. King Senusret I built a temple to Hathor, the goddess of turquoise, at the site that lasted about 800 years.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-posts/main/docs/images/alphabet/serabit-map.jpg" alt="Serabit el-Khadim" width="400">

In 1905, a husband and wife team of archaeologists Sir Flinders and Hilda Petrie were excavating at the site.
They noticed signs scratched on tall inscribed columns around the temple, on walls of the mines and by roads
that looked like simple imitations of hieroglyphs.
There also were a limited number of different signs, only a few dozen,
and they realized the Canaanite workers who could not read hieroglyphs were using thema as letters of an alphabet.
They took back a number of examples of the script to London including a
[small sandstone sphinx](https://earlysemitic.weebly.com/how-the-alphabet-was-born-from-hieroglyphs.html)
that had the letters on one side and proper hieroglyphics on the other.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-posts/main/docs/images/alphabet/serabit-sphinx-2.jpg" alt="Serabit sphinx 2" width="600">

In 1916 the Egyptologist Sir Alan Gardiner tested the idea that the inscriptions on both sides were the same,
that the sphinx was a Rosetta Stone for the new signs.
He guessed that the signs were based on the principle that they represented sounds,
which would be the first sound of the Semitic word for the symbol.
The hieroglyphic inscription was “The beloved of Hathor, the mistress of turquoise".
He recognized that the first of the signs was a house.
The Semitic word for house was bet, which began with the sound B.
The next sign was the head of an ox. The word in Semitic was alph, which began with the sound A.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-posts/main/docs/images/alphabet/pro-sinaic.jpg" alt="pro-sinaic.jpg" width="800">

He translated the rest of the signs as “The beloved of Baalat”, which is the Semitic term for the goddess Hathor.
He had made the first interpretation of the oldest found example of an alphabet.
The signs were dated to between 1850 and 1700 B.C.
The script was called Proto-Sinaitic and it was taken back to Canaan.
