# Origin of Data
The origin of data began when our ancestor organisms first
[sensed](https://colab.research.google.com/github/kushnertodd/basic-data-science-posts/blob/main/basic_data_science_data_history_senses.ipynb)
their environment to be able to change their behavior to survive better.

An ancestor developed an innate ability to immediately estimate quantities it could visually sense,
[number sense](https://colab.research.google.com/github/kushnertodd/basic-data-science-posts/blob/main/basic_data_science_data_history_number_sense.ipynb),
that gave an evolutionary advantage for finding food, avoiding predators, and defending territory.

Ancient evidence of primitive counting has been found in
[caves](https://colab.research.google.com/github/kushnertodd/basic-data-science-posts/blob/main/basic_data_science_data_history_basic_data_science_data_history_cave_art_calendar.ipynb)
around the world.

In the Middle East, the advent of agriculture led to an advanced system of
[counting](https://colab.research.google.com/github/kushnertodd/basic-data-science-posts/blob/main/basic_data_science_data_history_tally_mark_counting.ipynb)
that led to the development of writing and alphabets.

Formal
[counting systems](https://colab.research.google.com/github/kushnertodd/basic_data_science_data_history_decimals_accounting.ipynb)
evolved in Egypt, Greece, and Rome that allowed recording arbitrary abstract numbers and integrated numbers into everyday life.
A major innovation in counting occurred in India that simplified calculation and revolutionized commercial arithmetic and accounting.

[Data](https://www.ibm.com/think/topics/data) consists of symbols for descriptive facts, figures, observations, and other useful information.
These can be electronically stored as numbers, letters, and text for automatic processing in the computer.
The purpose of data science is to inspect, clean, transform, and interpret raw data to support data-driven decisions.
