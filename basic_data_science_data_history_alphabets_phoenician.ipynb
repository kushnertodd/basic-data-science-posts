## The Phoenician Alphabet

Around 1200 B.C. the [Bronze Age Collapse](https://www.history.com/articles/bronze-age-collapse-causes) occurred and the civilizations that used complex hieroglyphic and cuneiform writing, Egypt and Mesopotamia, were no longer thriving. Smaller city-states emerged and began to develop local languages.
The script was taken back to Canaan.
The Phoenicians were a Semitic tribe of seafaring merchants who adopted the script by 1000 B.C. They interpreted the script with their own signs.

By 900 B.C. they had spread the script throughout the Mediterranean. The script became the basis of the Aramaic alphabet. Aramaic became Arabic, Hebrew, and all other languages of the region.

Vowels were not part of this original Phoenician alphabet. But Wilson and other scholars believe that the original 22 symbols used could be drawn by the masses, including maritime traders who lived in what is now Lebanon, Syria and Israel

Written from right to left and spread by Phoenician maritime merchants who occupied part of modern Lebanon, Syria and Israel, this consonantal alphabet—also known as an abjad—consisted of 22 symbols simple enough for ordinary traders to learn and draw, making its use much more accessible and widespread.
