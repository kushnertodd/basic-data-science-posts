## The Phoenician Alphabet

Around 1200 B.C. the [Bronze Age Collapse](https://www.history.com/articles/bronze-age-collapse-causes) occurred and the civilizations that used complex hieroglyphic and cuneiform writing, Egypt and Mesopotamia, were no longer thriving.
Smaller city-states emerged and began to develop local languages.
The [Phoenicians](https://www.metmuseum.org/essays/the-phoenicians-1500-300-b-c)
were a Semitic tribe of seafaring merchants who were master shipbuilders and navigators, and lived in what is now Lebanon, Syria and Israel
and who created independent cities and trading posts around the Mediterranean.
They adopted the script by 1050 B.C. with their own signs, using 22 letters only for consonants that were read right to left.

By 900 B.C. they had spread the script throughout the Mediterranean because it was so efficient for trade.
The script became the basis of the Aramaic alphabet, which became Arabic, Hebrew, and all other languages of the region.
