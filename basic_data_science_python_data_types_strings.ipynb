## Escape Sequences
A backslash `\` in a string "escapes" the normal way it reads and interprets the next character.
It gives it a new meaning as a special character, or to include formatting in a string.

These are the most common backslash escapes:

| Escape sequence | Name | Use |
| - | - | - |
| \n | newline (line break) | break text across multiple lines |
| \t | tab (indent) | indent text to a fixed next tab stop |
| \\' | single quote | Insert a single quote in a single quoted string |
| \” | double quote | Insert a double quote in a double quoted string |
| \\\ | backslash |  Insert a backslash in a single or double quoted string |

### Raw Strings
Raw strings ignore backslashes, allowing escape sequences to be included as-is in the string. They begin with `r` followed by a single or double quoted string. Between the starting and ending quotes, backslashes `\` are not treated as special.
- `r'...'`
- `r"..."`

````python
print(r'single quoted raw strings ignore backslashes \ in escape sequences like \n and \t')
````

````python
print(r"double quoted raw strings ignore backslashes \ in escape sequences like \n and \t")
````

Raw strings beginning with `r` followed by a triple quotes (single or double quotes) can include single and double quotes without escaping them.
- `r'''...'''`
- `r"""..."""`

````python
print(r'''triple single quoted raw strings ignore escape sequences and can include single quotes ' ''')
````

````python
print(r"""triple double quoted raw strings ignore escape sequences and can include double quotes " """)
````

### String Slices
String *slices* access sequences of characters between positions in square brackets separated by one or more colons `:`.
The ways to use slices are:
- *string*`[`*start*`:`*end* `]`
- *string*`[`*start*`:`*end* `:`*step*`]`

The characters are selected from the start position up to, but not including, the end position.


````python
# the first word
print("DON'T PANIC!"[0:6])
````

````python
# the second word
print("DON'T PANIC!"[6:12])
````

## Searching Strings
There are two functions to search strings for a substring:

| Function | Description |
| - | - |
| `find()` | Find the starting position of the first occurrence of a substring |
| `rfind()` | Find the starting position of the last occurrence of a substring |


### `find()` function
There are three ways to use `find()`:

| use | meaning |
| - | - |
| *string*.`find(`*substring*`)` | Find first position of *substring* |
| *string*.`find(`*substring*`, `*start*`)` |  Find first position of *substring* beginning at *start* |
| *string*.`find(`*substring*`, `*start*`, `*end*`)` |  Find first position of *substring* between *start* and *end* |

`find()` returns position -1 if *substring* is not found.

````python
# find the position of first PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC"))
````

````python
# find the position of third PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC", 28))
````

````python
# find the position of the second PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC", 16, 38))
````

````python
# the position of PANIC if not found
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC", 34))
````

- __Exercise__. find the word using negative positions.

### `rfind()` function
There are three ways to use `rfind()`:

| use | meaning |
| - | - |
| *string*.`rfind(`*substring*`)` | Find first position of *substring* from the end|
| *string*.`rfind(`*substring*`, `*start*`)` |  Find first position of *substring* from the end beginning at *start* |
| *string*.`rfind(`*substring*`, `*start*`, `*end*`)` |  Find first position of *substring* from the end between *start* and *end* |

`rfind()` returns position -1 if *substring* is not found.

````python
# find the position of rightmost PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC"))
````

````python
# find the position of second rightmost PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC", 14, 26))
````

````python
# the position of rightmost PANIC if not found
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC", 35))
````

## Trimming Strings

| Function | Description |
| - | - |
| `lstrip()` | The string with leading characters removed |
| `rstrip()` | The string with trailing characters removed |
| `strip()` | The string with leading and trailing characters removed |


### `lstrip()` function
There are two ways to use `lstrip()`:

| use | meaning |
| - | - |
| *string*.`lstrip()` | Remove leading whitespace (space and tabs) |
| *string*.`lstrip(`*characters*`)` |  Remove any of the given leading *characters* |

````python
# remove leading spaces
print("   DON'T PANIC!".lstrip())
````

````python
# remove leading "   www."
print("   www.google.com".lstrip(" w."))
````

## String Properties
There are two functions to evaluate string properties:

| Function | Description |
| - | - |
| `count()` | The number of times a substring occurs in a string |
| `len()` | The number of characters in a string\ |


### `count()` function
There are three ways to use `count()`:

| use | meaning |
| - | - |
| *string*.`count(`*substring*`)` | The number of times a substring occurs in a string |
| *string*.`count(`*substring*`, `*start*`)` |  The number of times a substring occurs in a string beginning at *start* |
| *string*.`count(`*substring*`, `*start*`, `*end*`)` |  The number of times a substring occurs in a string between *start* and *end* |

`count()` returns 0 if *substring* is not found.

````python
# count the instances of "PANIC!"
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".count("PANIC!"))
````

````python
# count the instances of "PANIC!" starting in the middle
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".count("PANIC!", 11))
````

````python
# count the last instance of "PANIC!"
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".count("PANIC!", 22, 39))
````

## Formatting Strings

| `zfill()` | The numeric string left filled with zeros in a string of a given width |


### `center()` function
There are two ways to use `center()`:

| use | meaning |
| - | - |
| *string*.`center(`*width*`)` | The string centered in a blank string of a given width |
| *string*.`center(`*width*`, `*fillchar*`)` |  The string centered in a string of a given width filled out with *fillchar* |

````python
print("DON'T PANIC!".center(20))
````

````python
print("DON'T PANIC!".center(20, '-'))
````

## Modifying Strings
These are four functions to modify strings:

| Function | Description |
| - | - |
| `removeprefix()` | The string with a leading substring removed |
| `removesuffix()` | The string with a trailing substring removed |
| `replace()` | The string with a substring replaced a given of times |
| `join()` | The string with characters separated with a given substring |

### *string*.`removeprefix(`*substring*`)` function
The string with a leading substring removed.

````python
# strings can have a fixed leading substring removed
print("So Long, and Thanks for All the Fish".removeprefix("So Long, and "))
````

### `replace()` function
There are two ways to use `replace()`:

| use | meaning |
| - | - |
| *string*.`replace(`*old*`, `*new*`)` | The string with all substrings *old* replaced by *new* |
| *string*.`replace(`*old*`, `*new*`, `*count*`)` |  The string a given number of substrings *old* replaced by *new* |

````python
# all string substrings can be replaced
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".replace("PANIC", "BE SCARED"))
````

````python
# a fixed number of string substrings can be replaced
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".replace("PANIC", "BE SCARED", 2))
````
