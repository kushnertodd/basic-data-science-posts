## The Roman Alphabet
The Etruscans lived in what is now north-west Italy starting around 900 B.C.
Around 800 B.C. settlers from the Greek island of Euboea brought their variation of the Greek alphabet to Cumae, the first Green colony in Italy.
The Etruscans adopted that version around 700 B.C. to become the [Etruscan alphabet](https://www.omniglot.com/writing/etruscan.htm#google_vignette).
The Etruscans ruled early Rome and they adapted their alphabet to create the Latin alphabet around 700 B.C.

, and it was adopted to create Cyrillic by 900 A.D. The letters evolved as the languages developed.
