## The Alphabet
The alphabet is one of the most remarkable inventions of human civilization.

## The Birth of the Alphabet

The earliest writing systems, Egyptian hieroglyphics and Mesopotamian cuneiform, evolved over five thousand years ago.
Reading or writing these required knowing hundreds of signs and was limited to a highly trained elite.
Instead of hundreds of signs, alphabets consist of only twenty to thirty letters that represent basic spoken sounds
that can make a countless number of words.
Alphabets made knowledge available to everyone, no matter how skilled, and they could record knowledge instead of handing it down orally.

It only needed to be invented once.
The common ancestor of alphabetic script
may have begun with Semitic laborers at an Egyptian turquoise mine and
was spread by Phoenician traders to become almost every language in the world today.
