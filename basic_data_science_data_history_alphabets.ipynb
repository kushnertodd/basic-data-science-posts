## The Alphabet
The alphabet is one of the most remarkable inventions of human civilization.

Writing was originally available only to highly trained scribes and readers. Alphabets made it available to everyone, no matter how skilled, and they could record knowledge instead of transferring it orally.

It only needed to be invented once. It may have begun with Semitic laborers at an Egyptian turquoise mine and spread to all languages in the Western world.

By using a small number of symbols associated with sounds to make a countless number of words it could adapt to any group of people, in any location, and at any time.

At the site of [Serabit el-Khadim](https://www.penn.museum/sites/expedition/the-alphabet/) located in the south Sinai Peninsula were mines for turquoise that was prized by the Egyptian court and allowed rulers to pass to the afterlife. Around 1900 B.C, the king Senusret I built a temple to Hathor, the goddess of turquoise, at the site that lasted about 800 years. Laborers from the Semitic land of Canaan were imported to work on the mines.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-posts/main/docs/images/alphabet/serabit-map.jpg" alt="Serabit el-Khadim" width="400">

In 1905, a husband and wife team of archaeologists Sir Flinders and Hilda Petrie were excavating at the site, and they noticed signs that looked like hieroglyphics but there was a problem. They were much cruder than hieroglyphics and were scratched on columns in the temple, on walls of the mines and by roads. There also were a limited number of different signs, only a few dozen, and they recognized they were the signs of an alphabet. They took back a number of examples of the script to London, including a [small sandstone sphinx](https://earlysemitic.weebly.com/how-the-alphabet-was-born-from-hieroglyphs.html) that had the letters on the side. There were also hieroglyphics on the other side of the statue.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-posts/main/docs/images/alphabet/serabit-sphinx-2.jpg" alt="Serabit sphinx 2" width="600">

In 1916 the Egyptologist Sir Alan Gardiner tested the idea that the inscriptions on both sides were the same, that the sphinx was a Rosetta Stone for the new signs. He guessed that signs were based on the principle that they represented sounds, which would be the first sound of the Semitic word for the symbol. The hieroglyphic inscription was “the beloved of Hathor”, goddess of turquoise. He recognized that second of the signs was a house. The Semitic word for house was bet, which began with the sound B.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-posts/main/docs/images/alphabet/bet.jpg" alt="bet" width="200">

The next sign was the head of an ox. The word in Semitic was alph, which began with the sound A.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-posts/main/docs/images/alphabet/ox.jpg" alt="ox" width="200">

He translated the rest of the signs as “Baalat”, which is the Semitic term for the goddess Hathor. He had made the first interpretation of the oldest found example of an alphabet.

The signs were dated to between 1850 and 1700 B.C. They were taken back to Canaan. The script was called Proto-Sinaitic. Around 1200 B.C. the [Bronze Age Collapse](https://www.history.com/articles/bronze-age-collapse-causes) occurred and the civilizations that used complex hieroglyphic and cuneiform writing, Egypt and Mesopotamia, were no longer thriving. Smaller city-states emerged and began to develop local languages. The Phoenicians were a Semitic tribe of seafaring merchants who adopted the script by 1000 B.C. They interpreted the script with their own signs.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-posts/main/docs/images/alphabet/phoenician-alphabet.jpg" alt="phoenician alphabet" width="800">

By 900 B.C. they had spread the script throughout the Mediterranean. The script became the basis of the Aramaic alphabet. Aramaic became Arabic, Hebrew, and all other languages of the region. The alphabet included only consonants. By 800 B.C. the Greeks adopted the alphabet and added vowels, becoming the basis of all modern alphabets. The early alphabets were read right to left, but the Greek alphabet was eventually read as we do today, left to right. The Romans adopted it to create the Latin alphabet around 700 B.C., and it was adopted to create Cyrillic by 900 A.D. The letters evolved as the languages developed.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-posts/main/docs/images/alphabet/evolving-alphabet.jpg" alt="evolving alphabet" width="600">

The alphabet changed the lives of people for thousands of years. It may have been created by illiterate miners in ancient Egypt who borrowed from the complex hieroglyphics used at the time. It led to the creation of all Western languages. Along with the development of counting and numbers, the alphabet is one of the most essential elements of data and the basis for Data Science.
