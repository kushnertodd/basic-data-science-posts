An ancestor developed an innate ability to immediately estimate quantities it could visually sense,
[number sense](https://colab.research.google.com/github/kushnertodd/basic-data-science-posts/blob/main/basic_data_science_data_history_number_sense.ipynb),
that gave an evolutionary advantage for finding food, avoiding predators, and defending territory.

Both subitizing (the immediate, exact recognition of small sets up to 4) and the Approximate Number System (the ability to estimate and compare large quantities) evolved hundreds of millions of years ago, long before the arrival of humans. [1](https://academic.oup.com/cercor/article/doi/10.1093/cercor/bhaf021/8008571), [2](https://en.wikipedia.org/wiki/Subitizing), [3](https://www.smartick.com/blog/mathematics/number-sense-from-quantities-to-numbers/), [4](https://pmc.ncbi.nlm.nih.gov/articles/PMC13100464/)
Because these two mechanisms—collectively known as non-symbolic "number sense"—are shared by humans, monkeys, birds, fish, and even insects, scientists trace their evolutionary origins back to at least the common ancestor of all vertebrates (roughly 450 to 500 million years ago), with parallel mechanisms possibly developing even earlier in invertebrates. [1](https://royalsocietypublishing.org/rstb/article/373/1740/20160511/23389/Understanding-the-origin-of-number-sense-a-review), [2](https://pmc.ncbi.nlm.nih.gov/articles/PMC9253733/), [3](https://pubmed.ncbi.nlm.nih.gov/33926813/)

🧠 Why Did They Evolve So Early?

These systems did not evolve for "math" in the human sense. They evolved as critical survival advantages long before symbols, language, or digits existed: [1](https://iro.uiowa.edu/esploro/outputs/acceptedManuscript/Exploring-the-evolutionary-pathways-from-number/9984215256102771)
• Foraging and Feeding: Animals needed to instantly decide which patch of land had more food items or which hunting pack had better odds.
• Predator Avoidance: A fish or a bird needs to look at two social groups (shoals or flocks) and immediately merge with the larger group for safety in numbers.
• Territorial Defense: Animals like lions or chimpanzees track the number of roaring rivals versus the number of members in their own pack to decide whether to fight or retreat. [1](https://pages.ucsd.edu/~rnunez/COGS252_Readings/Nunez_TICS_2017%20copy.pdf), [2](https://royalsocietypublishing.org/rstb/article/373/1740/20160511/23389/Understanding-the-origin-of-number-sense-a-review)

🧬 Divergent vs. Convergent Evolution

There is an ongoing debate in cognitive biology about exactly how these systems spread: [1](https://pmc.ncbi.nlm.nih.gov/articles/PMC9253733/)
1. The Conserved Ancestor Theory: The core neural structures used for number sense (like the pallium in birds/fish and the parietal cortex in mammals) indicate that a primitive "number sense" was hardwired into the very first ancient vertebrates. [1](https://pmc.ncbi.nlm.nih.gov/articles/PMC2225440/), [2](https://pubmed.ncbi.nlm.nih.gov/33926813/), [3](https://royalsocietypublishing.org/rstb/article/373/1740/20160511/23389/Understanding-the-origin-of-number-sense-a-review)
2. The Convergent Evolution Theory: Because insects (like bees) can also track quantities despite having radically different brain architectures, it is highly likely that basic number sense is so fundamentally useful that nature evolved it independently multiple times. [1](https://pubmed.ncbi.nlm.nih.gov/33926813/), [2](https://pmc.ncbi.nlm.nih.gov/articles/PMC9253733/)
Summary: While exact symbolic mathematics (like writing "7" or "100") is a cultural invention only a few thousand years old, the biological engine behind it—subitizing and the ANS—is nearly half a billion years old.
