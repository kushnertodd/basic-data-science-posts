## The Alphabet in the Middle Ages
1. How Charlemagne’s Carolingian Lowercase Saved the Alphabet

Following the collapse of the Roman Empire, by the late 8th century many regions in Europe developed their own styles of writing the Latin alphabet.
These local scripts were becoming almost undecipherable outside the regions.

In 800 CE, when [Charlemagne](https://www.google.com/search?kgmid=%2Fm%2F01mhc&q=Charlemagne) became Holy Roman Emperor,
he realized he could not rule if people on one side of his empire could not read text from people on the other.
He commissioned the task of standardizing the script to make it clearer, resulting in a new script called *Carolingian minuscule*.

- Previously only uppercase letters were used, what would become our modern lowercase letters were added.
- Letters were given "ascenders" (like b or d) and "descenders" (like p or q) to make them more legible.
- The beginnings of punctuation to aid in reading were added, including symbols similar to periods, colons, semicolons, and question marks.
- Word spaces were added -- previously words ran into each other, making reading slow-going.
- Line breaks were made clearer and capital letters were added to begin lines.

These changes saved the Latin alphabet from fragmenting into incompatible regional scripts and made it the norm for reading across Europe.

Charlemagne’s influence was substantial and long-lasting,
A man obsessed with bringing order to his expanding kingdom, he sought reform in just about every sphere. For our story his most important contribution concerns his efforts to reform writing. Though efforts were already under way, he gave the job to a Yorkshire-man, Alcuin of York. Alcuin strove for clarity and uniformity. These efforts, with the backing of Charlemagne and the Church, brought about the Carolingian minuscule (or Carolingian script).

With Charlemagne and the Church behind it, the Carolingian script quickly spread across Europe, deposing a multitude of regional scripts on its way. By the second half of the tenth century, Carolingian script had reached England, replacing late forms of the Insular script; in Spain it replaced Visigothic.
